```python
from typing import Dict
import math
```

```python
# Dictionnaire des fréquences (classé par ordre décroissant), tiré de https://en.wikipedia.org/wiki/Letter_frequency
frequences:Dict[str, float] = {
    "E": 12.7,
    "T": 9.1,
    "A": 8.2,
    "O": 7.5,
    "I": 7.0,
    "N": 6.7,
    "S": 6.3,
    "H": 6.1,
    "R": 6.0,
    "D": 4.3,
    "L": 4.0,
    "C": 2.8,
    "U": 2.8,
    "M": 2.4,
    "W": 2.4,
    "F": 2.2,
    "G": 2.0,
    "Y": 2.0,
    "P": 1.9,
    "B": 1.5,
    "V": 0.98,
    "K": 0.77,
    "J": 0.16,
    "X": 0.15,
    "Q": 0.12,
    "Z": 0.074,
}
```

```python
# Boucle permettant de calculer l'entropie
entropy:float = 0
for freq in frequences.values(): # itération sur la fréquence d'apparition de chaque lettre
    proba:float = freq/100
    entropy += proba*math.log2(1/proba)
print(f"Entropie de l'anglais : {entropy}")
taille_initiale:float = math.log2(26)
compression = taille_initiale/entropy
print(f"Facteur de compression de l'anglais : {compression}")
```
